```python
catalog = "workspace"
schema_name = "default"

bronze_table = f"`{catalog}`.`{schema_name}`.`orders_bronze`"
silver_table = f"`{catalog}`.`{schema_name}`.`orders_silver`"

bronze_df = spark.table(bronze_table)

silver_df = (
    bronze_df
    .dropDuplicates(["order_id"])
    .dropna(subset=["amount"])
)

silver_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable(silver_table)

assert spark.table(silver_table).count() == 5

print("Silver completed successfully")
```
