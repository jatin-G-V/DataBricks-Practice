```python
from pyspark.sql.types import (
    StructType, StructField,
    IntegerType, StringType, DoubleType
)

# Change these to your practice catalog and schema
catalog = "workspace"
schema_name = "default"

orders_data = [
    (101, "Rahul", "Laptop", 50000.0, "Delivered"),
    (102, "Aman", "Mouse", 1000.0, "Delivered"),
    (103, "Priya", "Keyboard", 2000.0, "Cancelled"),
    (104, "Rahul", "Monitor", 12000.0, "Delivered"),
    (105, "Aman", "USB Cable", None, "Delivered"),
    (106, "Priya", "Headset", 3000.0, "Delivered"),
    (106, "Priya", "Headset", 3000.0, "Delivered")
]

schema = StructType([
    StructField("order_id", IntegerType()),
    StructField("customer", StringType()),
    StructField("product", StringType()),
    StructField("amount", DoubleType()),
    StructField("status", StringType())
])

df = spark.createDataFrame(orders_data, schema)

table_name = f"`{catalog}`.`{schema_name}`.`orders_bronze`"

df.write.format("delta").mode("overwrite").saveAsTable(table_name)

assert spark.table(table_name).count() == 7

print("Bronze completed successfully")
```
